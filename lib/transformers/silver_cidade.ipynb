```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
def transformSilverCidade(df, spark):
    # Exemplo de transformação para tabela 'cidades'
    df_tratado = ( df
          .withColumn("data_inclusao", to_timestamp(col("data_inclusao"), "yyyy-MM-dd HH:mm:ss"))
          .withColumn("data_atualizacao", to_timestamp(col("data_atualizacao"), "yyyy-MM-dd HH:mm:ss"))
          .withColumn("camada_origem", lit("bronze"))
          .withColumn("data_tratamento", current_timestamp())
    )
 
    return df_tratado
```
